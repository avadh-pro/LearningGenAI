# 🦜 LangChain Setup & Simple Chain — Explained

**What this notebook does:** the smallest possible LangChain example — installs the libraries, sets up a Cohere chat model behind a reusable prompt template, wraps them into an `LLMChain`, and runs it once to generate a joke. Each cell below has a short explanation directly above it.

### 📦 Cell 0 — Install the libraries

Installs LangChain's core package along with Cohere's SDK and the `langchain-cohere` integration (to use Cohere's chat models), `langchain_community` (community-maintained integrations), and `langchain-classic` (older APIs like `LLMChain`, which this notebook uses).

### 🧰 Cell 1 — Import the building blocks

`LLMChain` is the classic LangChain component that combines a prompt template with a model into one callable chain — this is the **Chains** component from the fundamentals notes. `PromptTemplate` lets you write a prompt with placeholders that get filled in later. `ChatCohere` is LangChain's wrapper for Cohere's chat models — the **Models** component doing the actual generation.

```python
from langchain_classic.chains import LLMChain
from langchain_core.prompts import PromptTemplate
from langchain_cohere import ChatCohere
```

### 🔑 Cell 2 — Import Colab's secret storage

`userdata` is Google Colab's way of safely storing API keys — like a Cohere API key — so they never get hard-coded or accidentally shared inside the notebook itself. Cell 3 reads the actual key back out with `userdata.get(...)`.

```python
from google.colab import userdata
```

### 🔗 Cell 3 — Build the prompt template and the chain

`prompt_template` is a string with a placeholder, `{adjective}` — this is what makes it *dynamic*: the same template can produce "Tell me a funny joke" or "Tell me a dark joke" just by changing one word. `PromptTemplate` wraps that string and tells LangChain which variable (`adjective`) needs to be filled in later. `LLMChain` then combines that prompt with a Cohere chat model into a single, reusable chain — the **Chains + Models + Prompts** components from the fundamentals video, wired together in three lines. `verbose=True` makes the chain print out exactly what prompt it sends to the model, useful for debugging.

**Example:** if `adjective` were filled in with `"dark"` instead of `"funny"`, this exact same chain would ask the model for a *dark* joke instead — no code changes needed, just a different input at run time.

```python
prompt_template = "Tell me a {adjective} joke"

prompt = PromptTemplate(
    input_variables=["adjective"], template=prompt_template
)

llm = LLMChain(llm=ChatCohere(cohere_api_key=userdata.get('COHERE_KEY'),model="command-a-03-2025"), prompt=prompt, verbose=True)
```

### ▶️ Cell 4 — Run the chain

`.invoke(...)` runs the whole chain end to end: it fills `{adjective}` in the prompt template with `"funny"`, producing "Tell me a funny joke", sends that to the Cohere model, and returns the result. Note the chain is still named `llm` here even though it's really a full chain (prompt + model), not just the raw model by itself.

```python
response = llm.invoke("funny")
```

### 🖨️ Cell 5 — Print just the generated text

`LLMChain`'s response comes back as a dictionary with a few fields; `response['text']` pulls out just the model's actual joke, rather than printing the whole raw response object.

```python
print(response['text'])
```
